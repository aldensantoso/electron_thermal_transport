```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("/Users/aldensantoso/Documents/plasma_research/ML Project").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import torch

from pinns.config import TRAIN_CSV
from pinns.data import load_training_data
from pinns.model import PINN
from pinns.train import train_pinn

df = load_training_data(TRAIN_CSV)

features = df[["P_beam", "B_T", "I_P", "n_e"]].values
target = df[["tau_e"]].values

X = torch.tensor(features, dtype=torch.float32)
y = torch.tensor(target, dtype=torch.float32)

model = PINN(input_dim=4, output_dim=1)
model = train_pinn(model, X, y, epochs=200, lr=1e-3)
```
